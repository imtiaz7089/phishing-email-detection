```python
import pandas as pd
import os

# Load the two datasets
spam = pd.read_csv(os.path.join(path, "SpamAssasin.csv"))
ceas = pd.read_csv(os.path.join(path, "CEAS_08.csv"))

# Fill missing text fields with empty text
for df in [spam, ceas]:
    df["subject"] = df["subject"].fillna("")
    df["body"] = df["body"].fillna("")
    df["receiver"] = df["receiver"].fillna("")

# Combine subject and body into one text field
for df in [spam, ceas]:
    df["email_text"] = (
        df["subject"].astype(str) + " " + df["body"].astype(str)
    )

print("SpamAssasin shape:", spam.shape)
print("CEAS_08 shape:", ceas.shape)

print("\nMissing values after cleaning:")
print("SpamAssasin:", spam.isnull().sum().sum())
print("CEAS_08:", ceas.isnull().sum().sum())
```

```python
print("SpamAssasin Labels:")
print(spam["label"].value_counts())
print("\nCEAS_08 Labels:")
print(ceas["label"].value_counts())

print("\nSample SpamAssasin:")
print(spam[["subject", "body", "label"]].head(3))

print("\nSample CEAS_08:")
print(ceas[["subject", "body", "label"]].head(3))
```

```python
# CEAS_08 dataset-এর প্রয়োজনীয় features তৈরি

df = ceas.copy()

# Subject + Body একসাথে
df["email_text"] = (
    df["subject"].astype(str) + " " + df["body"].astype(str)
)

# Sender থেকে domain বের করা
df["sender_domain"] = (
    df["sender"]
    .astype(str)
    .str.extract(r'@([^>\s]+)', expand=False)
    .fillna("")
)

# URL count
df["url_count"] = (
    df["urls"]
    .astype(str)
    .str.count(r'https?://|www\.')
)

print("Dataset shape:", df.shape)

print("\nFeatures:")
print(df[[
    "sender",
    "sender_domain",
    "subject",
    "email_text",
    "url_count",
    "label"
]].head())

print("\nLabel distribution:")
print(df["label"].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF Vectorizer
vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words="english"
)

X_text = vectorizer.fit_transform(df["email_text"])
y = df["label"]

print("TF-IDF shape:", X_text.shape)
print("Number of labels:", len(y))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Training labels:", y_train.shape[0])
print("Testing labels:", y_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Model তৈরি
model_lr = LogisticRegression(max_iter=1000)

# Training
model_lr.fit(X_train, y_train)

# Prediction
y_pred_lr = model_lr.predict(X_test)

# Performance
accuracy = accuracy_score(y_test, y_pred_lr)
precision = precision_score(y_test, y_pred_lr)
recall = recall_score(y_test, y_pred_lr)
f1 = f1_score(y_test, y_pred_lr)

print("Logistic Regression Results")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1-score :", round(f1, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_lr)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("\nTrue Negative (TN):", tn)
print("False Positive (FP):", fp)
print("False Negative (FN):", fn)
print("True Positive (TP):", tp)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Random Forest Model
model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Training
model_rf.fit(X_train, y_train)

# Prediction
y_pred_rf = model_rf.predict(X_test)

# Performance
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("Random Forest Results")
print("Accuracy :", round(accuracy_rf, 4))
print("Precision:", round(precision_rf, 4))
print("Recall   :", round(recall_rf, 4))
print("F1-score :", round(f1_rf, 4))
```

```python
import pandas as pd
import os

# Load datasets
nazario = pd.read_csv(os.path.join(path, "Nazario.csv"))
nigerian = pd.read_csv(os.path.join(path, "Nigerian_Fraud.csv"))
enron = pd.read_csv(os.path.join(path, "Enron.csv"))
ling = pd.read_csv(os.path.join(path, "Ling.csv"))

# Phishing emails
phishing = pd.concat(
    [
        nazario[["subject", "body"]],
        nigerian[["subject", "body"]]
    ],
    ignore_index=True
)

phishing["target"] = 1
phishing["source"] = "Phishing"

# Legitimate emails
legitimate = pd.concat(
    [
        enron[enron["label"] == 0][["subject", "body"]],
        ling[ling["label"] == 0][["subject", "body"]]
    ],
    ignore_index=True
)

legitimate["target"] = 0
legitimate["source"] = "Legitimate"

# Combine both classes
final_df = pd.concat(
    [phishing, legitimate],
    ignore_index=True
)

# Fill missing values
final_df["subject"] = final_df["subject"].fillna("")
final_df["body"] = final_df["body"].fillna("")

# Create combined email text
final_df["email_text"] = (
    final_df["subject"].astype(str)
    + " "
    + final_df["body"].astype(str)
)

# Remove duplicate emails
final_df = final_df.drop_duplicates(
    subset=["email_text"]
).reset_index(drop=True)

# Display results
print("Final dataset shape:", final_df.shape)

print("\nClass distribution:")
print(final_df["target"].value_counts())

print("\nSource distribution:")
print(final_df["source"].value_counts())

print("\nSample:")
print(final_df[["subject", "target", "source"]].head(10))
```

```python
# Balance the dataset

phishing_df = final_df[final_df["target"] == 1]
legitimate_df = final_df[final_df["target"] == 0]

# Use the same number of legitimate emails as phishing emails
legitimate_sample = legitimate_df.sample(
    n=len(phishing_df),
    random_state=42
)

balanced_df = pd.concat(
    [phishing_df, legitimate_sample],
    ignore_index=True
)

# Shuffle the dataset
balanced_df = balanced_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Balanced dataset shape:", balanced_df.shape)

print("\nClass distribution:")
print(balanced_df["target"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_text = balanced_df["email_text"]
y = balanced_df["target"]

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words="english"
)

# Fit only on training data
X_train = vectorizer.fit_transform(X_train_text)

# Transform test data
X_test = vectorizer.transform(X_test_text)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),
    "Multinomial Naive Bayes": MultinomialNB()
}

results = {}

for name, model in models.items():

    print(f"\nTraining {name}...")

    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    cm = confusion_matrix(y_test, y_pred)

    results[name] = {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "Confusion Matrix": cm
    }

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-score:  {f1:.4f}")
    print("Confusion Matrix:")
    print(cm)
```
