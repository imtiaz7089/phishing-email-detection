```python
import pandas as pd
import os

# Load the two datasets
spam = pd.read_csv(os.path.join(path, "SpamAssasin.csv"))
ceas = pd.read_csv(os.path.join(path, "CEAS_08.csv"))

# Fill missing text fields with empty text
for df in [spam, ceas]:
    df["subject"] = df["subject"].fillna("")
    df["body"] = df["body"].fillna("")
    df["receiver"] = df["receiver"].fillna("")

# Combine subject and body into one text field
for df in [spam, ceas]:
    df["email_text"] = (
        df["subject"].astype(str) + " " + df["body"].astype(str)
    )

print("SpamAssasin shape:", spam.shape)
print("CEAS_08 shape:", ceas.shape)

print("\nMissing values after cleaning:")
print("SpamAssasin:", spam.isnull().sum().sum())
print("CEAS_08:", ceas.isnull().sum().sum())
```

```python
print("SpamAssasin Labels:")
print(spam["label"].value_counts())
print("\nCEAS_08 Labels:")
print(ceas["label"].value_counts())

print("\nSample SpamAssasin:")
print(spam[["subject", "body", "label"]].head(3))

print("\nSample CEAS_08:")
print(ceas[["subject", "body", "label"]].head(3))
```

```python
# CEAS_08 dataset-এর প্রয়োজনীয় features তৈরি

df = ceas.copy()

# Subject + Body একসাথে
df["email_text"] = (
    df["subject"].astype(str) + " " + df["body"].astype(str)
)

# Sender থেকে domain বের করা
df["sender_domain"] = (
    df["sender"]
    .astype(str)
    .str.extract(r'@([^>\s]+)', expand=False)
    .fillna("")
)

# URL count
df["url_count"] = (
    df["urls"]
    .astype(str)
    .str.count(r'https?://|www\.')
)

print("Dataset shape:", df.shape)

print("\nFeatures:")
print(df[[
    "sender",
    "sender_domain",
    "subject",
    "email_text",
    "url_count",
    "label"
]].head())

print("\nLabel distribution:")
print(df["label"].value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF Vectorizer
vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words="english"
)

X_text = vectorizer.fit_transform(df["email_text"])
y = df["label"]

print("TF-IDF shape:", X_text.shape)
print("Number of labels:", len(y))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
print("Training labels:", y_train.shape[0])
print("Testing labels:", y_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Model তৈরি
model_lr = LogisticRegression(max_iter=1000)

# Training
model_lr.fit(X_train, y_train)

# Prediction
y_pred_lr = model_lr.predict(X_test)

# Performance
accuracy = accuracy_score(y_test, y_pred_lr)
precision = precision_score(y_test, y_pred_lr)
recall = recall_score(y_test, y_pred_lr)
f1 = f1_score(y_test, y_pred_lr)

print("Logistic Regression Results")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1-score :", round(f1, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_lr)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("\nTrue Negative (TN):", tn)
print("False Positive (FP):", fp)
print("False Negative (FN):", fn)
print("True Positive (TP):", tp)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Random Forest Model
model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Training
model_rf.fit(X_train, y_train)

# Prediction
y_pred_rf = model_rf.predict(X_test)

# Performance
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("Random Forest Results")
print("Accuracy :", round(accuracy_rf, 4))
print("Precision:", round(precision_rf, 4))
print("Recall   :", round(recall_rf, 4))
print("F1-score :", round(f1_rf, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

print("Random Forest Confusion Matrix:")
print(cm_rf)

tn_rf, fp_rf, fn_rf, tp_rf = cm_rf.ravel()

print("\nTrue Negative (TN):", tn_rf)
print("False Positive (FP):", fp_rf)
print("False Negative (FN):", fn_rf)
print("True Positive (TP):", tp_rf)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Naive Bayes Model
model_nb = MultinomialNB()

# Training
model_nb.fit(X_train, y_train)

# Prediction
y_pred_nb = model_nb.predict(X_test)

# Performance
accuracy_nb = accuracy_score(y_test, y_pred_nb)
precision_nb = precision_score(y_test, y_pred_nb)
recall_nb = recall_score(y_test, y_pred_nb)
f1_nb = f1_score(y_test, y_pred_nb)

print("Naive Bayes Results")
print("Accuracy :", round(accuracy_nb, 4))
print("Precision:", round(precision_nb, 4))
print("Recall   :", round(recall_nb, 4))
print("F1-score :", round(f1_nb, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm_nb = confusion_matrix(y_test, y_pred_nb)

print("Naive Bayes Confusion Matrix:")
print(cm_nb)

tn_nb, fp_nb, fn_nb, tp_nb = cm_nb.ravel()

print("\nTrue Negative (TN):", tn_nb)
print("False Positive (FP):", fp_nb)
print("False Negative (FN):", fn_nb)
print("True Positive (TP):", tp_nb)
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "Naive Bayes"
    ],
    "Accuracy": [
        accuracy,
        accuracy_rf,
        accuracy_nb
    ],
    "Precision": [
        precision,
        precision_rf,
        precision_nb
    ],
    "Recall": [
        recall,
        recall_rf,
        recall_nb
    ],
    "F1-score": [
        f1,
        f1_rf,
        f1_nb
    ]
})

print(results.to_string(index=False))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf, target_names=["Legitimate", "Phishing"]))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    model_rf,
    X_text,
    y,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

print("5-Fold Cross-Validation F1-scores:")
print(cv_scores)

print("\nMean F1-score:", round(cv_scores.mean(), 4))
print("Standard Deviation:", round(cv_scores.std(), 4))
```
